```python
students =[]

def add_students():
  student_id= input("enter student id : ")

  for student in students :
    if student["id"] == student_id:
      print ( " student id already exists! \n")
      return

  name = input(" enter student name : ")
  age = int(input("enter student's age : "))
  course = input("enter student's course: ")
  marks = float(input ("enter student's marks: "))

  student = {
      "id" : student_id,
      "Name" : name,
      "Age" : age,
      "Course" : course,
      "Marks" : marks
  }

  students.append(student)
  print(" student's data has been added successfully!!! ")


def view_students():
  if not students:
    print(" there are no student records to be found . ")


  print ("\n----------------records of the student------------")
  for student in students :
    print(f" ID : {student['id']}")
    print(f"Name : {student['Name']}")
    print(f"Age : {student["Age"]}")
    print(f"Course : {student["Course"]}")
    print(f"Marks : {student["Marks"]}")
    print("-"*40)

def search_student():
  value = input(" enter the id of the student id or student name to be searched").lower()

  found = False

  for student in students :
    if(student["id"].lower() == value or student["Name"].lower()) :
      print("\n student hs been found ")

      print(f" ID : {student['id']}")
      print(f"Name : {student['Name']}")
      print(f"Age : {student["Age"]}")
      print(f"Course : {student["Course"]}")
      print(f"Marks : {student["Marks"]}")

      found = True

  if not found :
    print("given student record doesn't exist ")


def update_student():
  value = input("enter the student id that needs to be updated ")

  for student in students :
    if (student["id"]==value ):
      print("\n eneter the new student details ")

      student["Name"] = input("enter the name of student: ")
      student["Age"] = input("enter the Age of the student: ")
      student["Course"] = input("enter the course of the student: ")
      student["Marks"] = input("enter the marks of the student : ")

      print("student data updated")

      return

  print("no student found ")


def delete_student():
  value = input("enter the student id tht needs to be deleted ")

  for student in students:
    if(student["id"] == value ):
      students.remove(student)
      print(f"student of id {value} deleted ")
      return

  print("student not found ")


while True:
  print("\n----------- student records menu----------")
  print("1. Add students")
  print("2. View students")
  print("3. Search students ")
  print("4. Update students ")
  print("5. delete students ")
  print("6. exit ")
  print("-"*40)

  choice = input("enter value between (1-6) : ")

  if choice == "1":
    add_students()

  elif choice == "2":
    view_students()

  elif choice == "3":
    search_student()

  elif choice == "4":
    update_student()

  elif choice == "5":
    delete_student()


  elif choice == "6":
    print(" thank you for using thr student's record menu ")
    break

  else :
    print("enter a valid choice between 1 to 6 ")







```
