```python
import csv
import os
from datetime import datetime

FILENAME = " expense_tracker.csv"

def create_csv():
  if not os.path.exists(FILENAME):
    with open(FILENAME,"w",newline="") as f:
      writer = csv.writer(f)
      writer.writerow(["Date" , "Category","Amount", "Note" ])


def add_expense():
  print("\n ------------ add your expense ----------------")

  while True:
    date = input("enter the date(dd-mm-yyyy)")

    try:
      datetime.strptime(date , "%d-%m-%Y")
      break
    except ValueError:
      print("invalid date time format , enter in the form dd-mm-yyyy")

  category = input("enter the category name: ").strip()

  if not category :
    print("category cannot be empty")
    return

  while True:
    try:
      amount = float(input("enter the amount: "))

      if amount<0:
        print("amount cannot be less than 0")
      else:
        break
    except ValueError:
      print("invalid . enter valid amount")

  note = input("enter note (optional) : ").strip()

  try:
      with open(FILENAME,"a",newline="") as f:
        writer =csv.writer(f)
        writer.writerow([date , category , amount , note])

      print("expenses have been added successfully ")
  except IOError:
    print("saving the expense failed")


def view_expense():
  print("\n -------------- viewing the expenses--------------")

  try:
    with open(FILENAME,"r",newline="") as f:
      reader = csv.DictReader(f)

      expenses= list(reader)

      if not expenses:
        print("there is no recorded expense")
        return

      total= 0

      print(f"{'Date:<15'},{'Category :<15'} , {'Amount': <15} , {'Note'}")

      print("="*40)

      for expense in expenses:
        amount = float(expense['Amount'])
        total+=amount


        print(
            f"{expense['Date']:<15}"
            f"{expense['Category']:<15}"
            f"{expense['Amount']:<15}"
            f"{expense['Note']}"
        )

      print("="*40)
      print(f"total amount : {total:.2f}")
  except FileNotFoundError:
    print("no such file found")

  except (IOError, ValueError):
    print("an error has occured while reading data")


def category_summary():
  print("\n--------- summary of each category------------")

  category_tot={}

  try:
    with open(FILENAME,"r",newline="") as f:
      reader = csv.DictReader(f)

      for expense in reader:
        category = expense['Category']
        amount = float(expense['Amount'])

        if category in category_tot:
          category_tot[category]+=amount
        else:
          category_tot[category] = amount

    if not category_tot:
      print("no expense records present")
      return

    for category, total in category_tot.items():
      print(f"{category} : {total:.2f}")

  except FileNotFoundError:
    print("no expenses file found ")

  except ( ValueError , IOError):
    print("an error has occured while reading data")


def main():
    create_csv()

    while True:
        print("\n===================================")
        print("       EXPENSE TRACKER")
        print("===================================")
        print("1. Add Expense")
        print("2. View All Expenses")
        print("3. Category-wise Spending Summary")
        print("4. Exit")
        print("===================================")

        choice = input("Enter your choice (1-4): ")

        if choice == "1":
            add_expense()

        elif choice == "2":
            view_expense()

        elif choice == "3":
            category_summary()

        elif choice == "4":
            print("Thank you for using Expense Tracker!")
            break

        else:
            print("Invalid choice! Please enter a number between 1 and 4.")


if __name__ == "__main__":
    main()
```
