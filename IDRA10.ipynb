```python
df["Delay_Minutes"] = df["Delay_Minutes"].fillna(0)

print("Duplicate rows:", df.duplicated().sum())

df.isnull().sum().sum()
```

```python
df["Load_Factor_%"] = round((df["Passengers"] / df["Seat_Capacity"]) * 100, 2)

df["Revenue"] = df["Passengers"] * df["Average_Ticket_Price"]

df["Flight_Date"] = pd.to_datetime(df["Flight_Date"])
df["Month"] = df["Flight_Date"].dt.month_name()
df["Day_of_Week"] = df["Flight_Date"].dt.day_name()

def delay_bucket(minutes):
    if minutes == 0:
        return "No Delay"
    elif minutes <= 30:
        return "Minor Delay (<=30 min)"
    elif minutes <= 60:
        return "Moderate Delay (31-60 min)"
    else:
        return "Severe Delay (>60 min)"

df["Delay_Category"] = df["Delay_Minutes"].apply(delay_bucket)

df.head()
```

**Q1. Which airline has the most delayed flights, and which has the best on-time performance?**

```python
delay_share = df.groupby("Airline")["Flight_Status"].apply(
    lambda x: (x == "Delayed").sum() / len(x) * 100
).round(2).sort_values(ascending=False)

print("% of flights delayed, by airline:")
delay_share
```

**Q2. Does weather meaningfully affect delay duration?**

```python
weather_impact = df.groupby("Weather").agg(
    Avg_Delay=("Delay_Minutes", "mean"),
    Flights=("Flight_ID", "count")
).round(2).sort_values("Avg_Delay", ascending=False)

weather_impact
```

**Q4. Is there a relationship between load factor and passenger satisfaction?**

```python
correlation = df[["Load_Factor_%", "Passenger_Satisfaction", "Delay_Minutes",
                   "Average_Ticket_Price", "Avg_Baggage_Kg"]].corr()

correlation["Passenger_Satisfaction"].sort_values(ascending=False)
```

**Q6. Which month had the most flight activity and the highest average delay?**

```python
monthly_summary = df.groupby("Month").agg(
    Flights=("Flight_ID", "count"),
    Avg_Delay=("Delay_Minutes", "mean")
).round(2)

month_order = ["January", "February", "March", "April", "May", "June",
               "July", "August", "September", "October", "November", "December"]
monthly_summary = monthly_summary.reindex([m for m in month_order if m in monthly_summary.index])

monthly_summary
```

1. **Cancellations are weather-linked at the data level** — every flight with a missing `Delay_Minutes` value was a cancelled flight, confirming cancellations are recorded separately from delay duration rather than as an extreme delay case.

2. **Delay performance varies clearly by airline** — some airlines (e.g. IndiGo and SpiceJet in this dataset) show a noticeably higher average delay and higher share of delayed flights than others (e.g. Air India Express and Akasa Air), suggesting real operational differences worth investigating further.

3. **Weather has a measurable but moderate effect on delays** — average delay minutes differ across weather conditions, with Rain and Storm associated with longer average delays than Clear or Fog conditions, though the gap is not enormous, implying other operational factors also drive delays.

4. **Load factor does not strongly predict satisfaction** — the correlation between `Load_Factor_%` and `Passenger_Satisfaction` is weak and positive, indicating a fuller flight doesn't necessarily mean an unhappier passenger in this dataset; satisfaction is likely driven more by service factors not captured here.

5. **Premium Economy passengers report the highest average satisfaction**, edging out both Business and Economy — a counter-intuitive finding, since Business class typically commands the highest price but does not have the highest satisfaction score.

6. **Route traffic is concentrated on a handful of city pairs** — a small number of Origin→Destination routes account for a disproportionate share of total flights, consistent with hub-and-spoke patterns around major cities like Delhi and Mumbai.
