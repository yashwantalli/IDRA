```python
n1=float(input("first num \n"))
n2=float(input("second num \n"))

add = n1+n2
sub = n1-n2
multi= n1*n2

print(f"Addition : {add}")
print(f"Subtraction : {sub}")
print(f"Multiplication : {multi}")

if n2 != 0:
  div = n1/n2
  floor_div = n1//n2
  mod = n1%n2

  print(f"Division : {div}")
  print(f"floor_div : {floor_div}")
  print(f"Modulus : {mod}")
else:
  print("Division, Floor_division, Modulus cannot be done by 'Zero' \n")

exp = n1**n2

print(f"Exponential : {exp}")
```
