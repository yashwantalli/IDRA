```python
import pandas as pd
import numpy as np
```

```python
raw_df= pd.read_csv("/content/Day11_Messy_Company_Employee_Dataset.csv")
df=raw_df.copy()
print(df.shape)
df.head()
```

```python
df.info()
```

```python
df.describe(include ="all")
```

```python
missing_summary = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percent": (df.isnull().sum() / len(df) * 100).round(2)
})
missing_summary = missing_summary[missing_summary["Missing_Count"] > 0].sort_values(
    "Missing_Count", ascending=False
)
missing_summary
```

```python
categorical_cols = ["Department", "Gender", "City", "Work_Mode"]

for col in categorical_cols:
    print(f"--- {col} ---")
    print(df[col].value_counts(dropna=False))
    print()
```

```python
whitespace_hits = df.loc[df["City"].notna() & (df["City"] != df["City"].str.strip()), "City"]
print("Values with leading/trailing whitespace:", whitespace_hits.tolist())
```

```python
print(df.dtypes)
print()
print("Sample Joining_Date values:", df['Joining_Date'].head(3).tolist(), "-> currently stored as:", df['Joining_Date'].dtype)
```

```python
full_duplicates = df.duplicated().sum()
id_duplicates = df["Employee_ID"].duplicated().sum()

print("Fully duplicated rows (every column matches):", full_duplicates)
print("Duplicate Employee_IDs:", id_duplicates)

df[df.duplicated(keep=False)].sort_values("Employee_ID")
```

```python
rows_before = len(df)
df = df.drop_duplicates(keep="first").reset_index(drop=True)
rows_after = len(df)

print(f"Rows before: {rows_before}")
print(f"Rows after:  {rows_after}")
print(f"Duplicate rows removed: {rows_before - rows_after}")
print("Remaining duplicate Employee_IDs:", df['Employee_ID'].duplicated().sum())
```

```python
text_cols = ["Employee_Name", "Department", "Job_Title", "Gender", "City", "Work_Mode"]

for col in text_cols:
    df[col] = df[col].str.strip()


df["Department"] = df["Department"].str.title()
df["City"] = df["City"].str.title()
df["Job_Title"] = df["Job_Title"].str.title()
df["Employee_Name"] = df["Employee_Name"].str.title()

df["Gender"] = df["Gender"].str.capitalize()
df["Work_Mode"] = df["Work_Mode"].str.capitalize()

print("Department after cleaning:", sorted(df["Department"].dropna().unique()))
print("Gender after cleaning:", sorted(df["Gender"].dropna().unique()))
print("City after cleaning:", sorted(df["City"].dropna().unique()))
print("Work_Mode after cleaning:", sorted(df["Work_Mode"].dropna().unique()))
```

```python
department_mode = df["Department"].mode()[0]
df["Department"] = df["Department"].fillna(department_mode)
print(f"Department filled with mode: '{department_mode}'")
```

```python
df["Gender"] = df["Gender"].fillna("Not Specified")
print("Gender missing values filled with 'Not Specified'")
print(df["Gender"].value_counts())
```

```python
missing_city_before = df["City"].isnull().sum()
df["City"] = df["City"].ffill()
print(f"City: {missing_city_before} missing values forward-filled")
print("Remaining missing City values:", df["City"].isnull().sum())
```

```python
work_mode_mode = df["Work_Mode"].mode()[0]
df["Work_Mode"] = df["Work_Mode"].fillna(work_mode_mode)
print(f"Work_Mode filled with mode: '{work_mode_mode}'")
```

```python
age_median = df["Age"].median()
df["Age"] = df["Age"].fillna(age_median)
print(f"Age filled with overall median: {age_median}")
```

```python
salary_before_missing = df["Annual_Salary"].isnull().sum()
df["Annual_Salary"] = df.groupby("Job_Title")["Annual_Salary"].transform(
    lambda s: s.fillna(s.median())
)
overall_salary_median = df["Annual_Salary"].median()
df["Annual_Salary"] = df["Annual_Salary"].fillna(overall_salary_median)

print(f"Annual_Salary: {salary_before_missing} missing values filled using each employee's Job_Title median")
print("Remaining missing Annual_Salary values:", df["Annual_Salary"].isnull().sum())
```

```python
experience_median = df["Experience_Years"].median()
df["Experience_Years"] = df["Experience_Years"].fillna(experience_median)
print(f"Experience_Years filled with overall median: {experience_median}")

performance_mode = df["Performance_Score"].mode()[0]
df["Performance_Score"] = df["Performance_Score"].fillna(performance_mode)
print(f"Performance_Score filled with mode: {performance_mode}")
```

```python
print("Remaining missing values after all imputation steps:")
print(df.isnull().sum().sum(), "total missing values left")
```

```python
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"], format="%Y-%m-%d")

df["Age"] = df["Age"].round().astype(int)
df["Performance_Score"] = df["Performance_Score"].round().astype(int)
df["Annual_Salary"] = df["Annual_Salary"].round().astype(int)

df["Experience_Years"] = df["Experience_Years"].round(1)

print(df.dtypes)
```

```python
comparison = pd.DataFrame({
    "Raw Dataset": [
        raw_df.shape[0],
        raw_df.shape[1],
        raw_df.isnull().sum().sum(),
        raw_df.duplicated().sum(),
        raw_df["Joining_Date"].dtype,
        raw_df["Department"].nunique(),
        raw_df["Gender"].nunique(),
        raw_df["City"].nunique(),
        raw_df["Work_Mode"].nunique(),
    ],
    "Cleaned Dataset": [
        df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        df["Joining_Date"].dtype,
        df["Department"].nunique(),
        df["Gender"].nunique(),
        df["City"].nunique(),
        df["Work_Mode"].nunique(),
    ],
}, index=[
    "Row Count", "Column Count", "Total Missing Values", "Duplicate Rows",
    "Joining_Date dtype", "Unique Department values", "Unique Gender values",
    "Unique City values", "Unique Work_Mode values",
])

comparison
```

```python
output_path = "Cleaned_Company_Employee_Dataset.csv"
df.to_csv(output_path, index=False)
print(f"Cleaned dataset exported to: {output_path}")
print(f"Final shape: {df.shape}")
```
