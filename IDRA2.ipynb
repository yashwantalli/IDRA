```python
import random

max_attempts = 8

secretnum = random.randint(1,100)

print("NUMBER GUESSING GAME \n")
print("The number ranges from 1 - 100 \n")
print(f"you have {max_attempts} to guess the number\n")

for attempt in range(1,max_attempts+1):
  guess = int(input(f"attempt no.{attempt} to guess the number."))

  if guess == secretnum:
    print(f" You have guessed the number right. it is {guess} and you have guessed it in {attempt} tries .\n")
    break

  elif guess< secretnum :
    print("the guessed number is too low . try again \n")

  else :
    print("the guessed number is too high . try again \n")
else :
  print(f" u have exhausted ur no.of attempts ({max_attempts})")
  print(f" the chosen number was {secretnum}")
```
