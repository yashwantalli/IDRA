```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/Day8_Ecommerce_Sales_Dataset.csv')

print("Shape:", df.shape)
df.head()
```

```python
orders_df = df[['Order_ID', 'Order_Date', 'Customer', 'City', 'Region',
                 'Quantity', 'Total_Sales', 'Payment_Method']].copy()

product_details_df = df[['Order_ID', 'Category', 'Product', 'Unit_Price',
                          'Discount_Percent', 'Rating']].copy()

print("orders_df:", orders_df.shape)
print("product_details_df:", product_details_df.shape)

orders_df.head()
```

```python
product_details_df.head()
```

```python
merged_df = pd.merge(orders_df, product_details_df, on='Order_ID', how='inner')

print("Shape after merge():", merged_df.shape)
assert merged_df.shape[0] == df.shape[0], "Row count changed after merge — check the join!"
merged_df.head()
```

```python
batch_north_east = merged_df[merged_df['Region'].isin(['North', 'East'])]
batch_south_west = merged_df[merged_df['Region'].isin(['South', 'West'])]

print("North & East batch:", batch_north_east.shape)
print("South & West batch:", batch_south_west.shape)

combined_df = pd.concat([batch_north_east, batch_south_west], axis=0, ignore_index=True)

print("Shape after concat():", combined_df.shape)
assert combined_df.shape[0] == merged_df.shape[0], "Row count mismatch after concat()!"
assert set(combined_df['Order_ID']) == set(merged_df['Order_ID']), "Order_IDs don't match after concat()!"

combined_df.head()
```

```python
combined_df['Order_Date'] = pd.to_datetime(combined_df['Order_Date'])

combined_df['Order_Year'] = combined_df['Order_Date'].dt.year
combined_df['Order_Month'] = combined_df['Order_Date'].dt.month
combined_df['Order_Month_Name'] = combined_df['Order_Date'].dt.month_name()
combined_df['Order_Day'] = combined_df['Order_Date'].dt.day
combined_df['Order_DayOfWeek'] = combined_df['Order_Date'].dt.day_name()
combined_df['Order_Quarter'] = combined_df['Order_Date'].dt.quarter
combined_df['Is_Weekend'] = combined_df['Order_Date'].dt.dayofweek.apply(lambda d: d >= 5)

combined_df[['Order_ID', 'Order_Date', 'Order_Year', 'Order_Month_Name',
             'Order_Day', 'Order_DayOfWeek', 'Order_Quarter', 'Is_Weekend']].head()
```

```python
combined_df['Discount_Amount'] = combined_df.apply(
    lambda row: round(row['Unit_Price'] * row['Quantity'] * row['Discount_Percent'] / 100, 2),
    axis=1
)

def price_category(price):
    if price < 1000:
        return 'Budget'
    elif price < 2500:
        return 'Mid-Range'
    else:
        return 'Premium'

combined_df['Price_Category'] = combined_df['Unit_Price'].apply(price_category)

def value_segment(total):
    if total < 2000:
        return 'Low Value'
    elif total < 6000:
        return 'Medium Value'
    else:
        return 'High Value'

combined_df['Customer_Value_Segment'] = combined_df['Total_Sales'].apply(value_segment)

def rating_label(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    elif rating >= 3.5:
        return 'Average'
    else:
        return 'Poor'

combined_df['Rating_Label'] = combined_df['Rating'].apply(rating_label)

combined_df[['Order_ID', 'Unit_Price', 'Price_Category', 'Discount_Amount',
             'Total_Sales', 'Customer_Value_Segment', 'Rating', 'Rating_Label']].head()
```

```python
final_columns = [
    'Order_ID', 'Order_Date', 'Order_Year', 'Order_Month', 'Order_Month_Name',
    'Order_Day', 'Order_DayOfWeek', 'Order_Quarter', 'Is_Weekend',
    'Customer', 'City', 'Region',
    'Category', 'Product', 'Price_Category',
    'Quantity', 'Unit_Price', 'Discount_Percent', 'Discount_Amount',
    'Total_Sales', 'Customer_Value_Segment',
    'Rating', 'Rating_Label', 'Payment_Method'
]

final_df = combined_df[final_columns].sort_values('Order_Date').reset_index(drop=True)

print("Final shape:", final_df.shape)
final_df.head(10)
```
